## 三种控制结构

程序有三种基本控制结构：

```
┌──────────┐    ┌──────────────┐    ┌──────────────┐
│  顺序     │    │   决策       │    │   循环       │
│ Sequence  │    │  Decision    │    │  Iteration   │
├──────────┤    ├──────────────┤    ├──────────────┤
│ 语句1    │    │    条件?     │    │    条件?     │
│   ↓      │    │   ↙    ↘    │    │   ↙    ↘    │
│ 语句2    │    │ True  False  │    │ True  False  │
│   ↓      │    │  ↓     ↓    │    │  ↓     ↓    │
│ 语句3    │    │语句A  语句B  │    │循环体  退出  │
└──────────┘    └──────────────┘    │  ↓         │
                                    │ 回到条件───┘
                                    └──────────────┘
```

## 第一部分：布尔值（Booleans）

### 什么是布尔值？

**布尔值**（Boolean）是计算机中最简单的数据类型，只有两个值：

| 值 | 含义 | 类比 |
|:---|:---|:---|
| `True` | 真 / 是 | 灯亮着 💡 |
| `False` | 假 / 否 | 灯灭了 🌑 |

就像开关只有"开"和"关"两种状态，布尔值也只有 `True` 和 `False`。

> ⚠️ 注意：`True` 和 `False` 首字母必须**大写**！

````python
# 布尔值的基本使用
a = True
b = False

print("a =", a, "  →  类型:", type(a))
print("b =", b, "  →  类型:", type(b))

# 直接查看 True 和 False 的类型
print("\nTrue 的类型:", type(True))
print("False 的类型:", type(False))
````

---

### 布尔值从哪里来？—— 比较运算符

布尔值最常见的来源是**比较运算**。你问计算机一个问题，它回答 `True` 或 `False`。

| 运算符 | 含义 | 示例 | 结果 |
|:---:|:---|:---|:---:|
| `==` | 等于？ | `3 == 5` | `False` |
| `!=` | 不等于？ | `3 != 5` | `True` |
| `>` | 大于？ | `3 > 5` | `False` |
| `<` | 小于？ | `3 < 5` | `True` |
| `>=` | 大于等于？ | `5 >= 5` | `True` |
| `<=` | 小于等于？ | `5 <= 3` | `False` |

> ⚠️ 特别注意：**等于号是 `==`（两个等号）**，一个 `=` 是赋值！

````python
# 比较运算符示例
a = 1
b = 2

print(f"a = {a}, b = {b}")
print("-" * 30)
print(f"a == b  →  {a == b}    # a 等于 b 吗？")
print(f"a != b  →  {a != b}    # a 不等于 b 吗？")
print(f"a > b   →  {a > b}     # a 大于 b 吗？")
print(f"a < b   →  {a < b}     # a 小于 b 吗？")
print(f"a >= b  →  {a >= b}    # a 大于等于 b 吗？")
print(f"a <= b  →  {a <= b}    # a 小于等于 b 吗？")
````

### 混合数值类型也可以比较

整数 `int` 和小数 `float` 之间可以互相比较：

````python
# 混合类型比较：int 和 float
print("1 == 2.0  →", 1 == 2.0)    # 整数 1 等于 小数 2.0 吗？
print("1 != 2.0  →", 1 != 2.0)    # 整数 1 不等于 小数 2.0 吗？
print("1.0 > 2   →", 1.0 > 2)     # 小数 1.0 大于 整数 2 吗？
print("1.0 < 2   →", 1.0 < 2)     # 小数 1.0 小于 整数 2 吗？
print("3.0 >= 3  →", 3.0 >= 3)    # 小数 3.0 大于等于 整数 3 吗？
print("2.5 <= 2  →", 2.5 <= 2)    # 小数 2.5 小于等于 整数 2 吗？
````

---

### 字符串也能比较？

字符串的比较是按**字典序（lexicographical order）**——也就是按 ASCII 码逐字符比较，类似查字典的顺序。

每个字符在计算机内部都有一个数字编号（ASCII 码）：

| 字符 | ASCII | 字符 | ASCII | 字符 | ASCII |
|:---:|:---:|:---:|:---:|:---:|:---:|
| `'0'` | 48 | `'A'` | 65 | `'a'` | 97 |
| `'1'` | 49 | `'B'` | 66 | `'b'` | 98 |
| ... | ... | `'Z'` | 90 | `'z'` | 122 |

比较时，从左到右逐个字符比较，遇到不同的字符就出结果。

````python
# 查看字符的 ASCII 码
print("ord('a') =", ord('a'))   # a 的 ASCII 码
print("ord('b') =", ord('b'))   # b 的 ASCII 码
print("ord('A') =", ord('A'))   # 大写 A 的 ASCII 码

# 字符串比较
print("\n字符串比较：")
print("'a' < 'b'     →", 'a' < 'b')        # 97 < 98 → True
print("'a' == 'a'    →", 'a' == 'a')       # 相同 → True
print("'aa' < 'ab'   →", 'aa' < 'ab')      # 前一个字符相同，比较第二个
print("'aba' > 'aab'  →", 'aba' > 'aab')   # 第二个字符 b > a → True
print("'apple' < 'banana' →", 'apple' < 'banana')  # a < b
````

### ⚠️ 字符串和数字不能比较大小！

如果拿字符串和数字比 `>` `<` `>=` `<=`，Python 会**报错**：

````python
# == 和 != 可以比较字符串和数字（但永远不相等）
print("'a' == 2  →", 'a' == 2)     # 永远是 False
print("'a' != 2  →", 'a' != 2)     # 永远是 True

# 但 <, >, <=, >= 会报错！
print("\n尝试 'a' < 2 ...")
try:
    print('a' < 2)
except TypeError as e:
    print(f"❌ 报错了！{e}")
    print("TypeError: 字符串和整数不能比较大小")
````

### 布尔值也是数字！

一个有趣的冷知识：`True` 等于 `1`，`False` 等于 `0`，它们可以直接参与数学运算！

````python
# 布尔值就是特殊的数字
print("True == 1  →", True == 1)       # True
print("False == 0 →", False == 0)      # True
print("True > False →", True > False)  # 1 > 0 → True

# 布尔值参与数学运算
print("\n数学运算：")
print("True + True   =", True + True)     # 1 + 1 = 2
print("True + False  =", True + False)    # 1 + 0 = 1
print("True * 100    =", True * 100)      # 1 * 100 = 100
print("False * 100   =", False * 100)     # 0 * 100 = 0
print("True + (False / True) =", True + (False / True))  # 1 + 0/1 = 1.0
````

### 🔍 成员运算符：`in` 和 `not in`

检查一个东西是否在另一个东西
：

| 运算符 | 含义 | 示例 |
|:---|:---|:---|
| `in` | 在…里面？ | `'a' in 'apple'` → `True` |
| `not in` | 不在…里面？ | `'x' not in 'apple'` → `True` |

````python
# 成员运算符
print("'a' in 'abc'      →", 'a' in 'abc')        # 'a' 在 'abc' 里吗？
print("'A' not in 'abc'  →", 'A' not in 'abc')    # 大小写敏感！
print("'ab' in 'abc'     →", 'ab' in 'abc')       # 子串也能检测
print("'ac' not in 'abc'  →", 'ac' not in 'abc')  # 'ac' 不是连续子串

# 也可以用于列表
fruits = ['apple', 'banana', 'cherry']
print("\n'apple' in fruits  →", 'apple' in fruits)
print("'grape' not in fruits →", 'grape' not in fruits)
````

### 🧠 逻辑运算符：`not`、`and`、`or`

逻辑运算符用来**组合多个条件**，就像把多个判断题串起来。

| 运算符 | 含义 | 说明 |
|:---|:---|:---|
| `not a` | 取反 | `a` 为 `True` 时返回 `False`，反之亦然 |
| `a and b` | 与（并且） | 两个都 `True` 才返回 `True` |
| `a or b` | 或（或者） | 至少一个 `True` 就返回 `True` |

````python
# 逻辑运算符示例
print("=== not（取反）===")
print("not True  →", not True)
print("not False →", not False)

print("\n=== and（与）===")
print("True and True   →", True and True)
print("True and False  →", True and False)
print("False and True  →", False and True)
print("False and False →", False and False)

print("\n=== or（或）===")
print("True or True   →", True or True)
print("True or False  →", True or False)
print("False or True  →", False or True)
print("False or False →", False or False)
````

### 📊 真值表（Truth Tables）

把上面的结果整理成表格，就是**真值表**：

#### `not`（取反）

| a | not a |
|:---:|:---:|
| False | **True** |
| True | **False** |

#### `and`（与）

| a | b | a and b |
|:---:|:---:|:---:|
| False | False | False |
| False | True | False |
| True | False | False |
| True | True | **True** |

#### `or`（或）

| a | b | a or b |
|:---:|:---:|:---:|
| False | False | False |
| False | True | **True** |
| True | False | **True** |
| True | True | **True** |

> 🎯 记忆口诀：
> - `and`：**全真才真**（像串联电路，有一个断开就全断）
> - `or`：**有真就真**（像并联电路，有一个接通就亮）

### 实际应用示例

把比较运算符和逻辑运算符组合起来，就能写出真正的判断逻辑：

````python
# 实际例子：判断一个学生是否符合条件
age = 15
grade = 85
is_student = True

# 条件：年龄 >= 12 且 成绩 >= 60 且 是学生
result = (age >= 12) and (grade >= 60) and is_student
print(f"年龄={age}, 成绩={grade}, 是学生={is_student}")
print(f"符合条件？ {result}")

# 条件：年龄 < 12 或 成绩 < 60
result2 = (age < 12) or (grade < 60)
print(f"\n需要特别关注？ {result2}")

# not 的使用
is_holiday = False
print(f"\n今天放假吗？ {is_holiday}")
print(f"今天上学吗？ {not is_holiday}")
````

### 📐 运算符优先级（Operator Precedence）

当表达式中有多种运算符混在一起，Python 按以下优先级从高到低计算：

| 优先级 | 运算符 | 说明 |
|:---:|:---|:---|
| 1 (最高) | `()` | 括号 — 优先计算 |
| 2 | `**` | 指数（幂运算） |
| 3 | `+x`, `-x` | 正号、负号 |
| 4 | `*`, `/`, `//`, `%` | 乘、除、整除、取余 |
| 5 | `+`, `-` | 加、减 |
| 6 | `==`, `!=`, `<`, `<=`, `>=`, `>` | 比较运算符 |
| 7 | `not` | 逻辑非 |
| 8 | `and` | 逻辑与 |
| 9 (最低) | `or` | 逻辑或 |

> 🎯 记不住优先级？**加括号！**括号里的先算，清晰又不会出错。

````python
# 运算符优先级示例
# 不加括号 — 容易搞混
a = True
b = False
c = True

# not a or not b and c
# 按优先级等价于: (not a) or ((not b) and c)
result_no_paren = not a or not b and c
result_with_paren = (not a) or ((not b) and c)

print(f"not a or not b and c         = {result_no_paren}")
print(f"(not a) or ((not b) and c)   = {result_with_paren}")
print(f"结果相同 ✓")
````

### 🔗 运算符链式求值

看一个复杂表达式是如何一步步算出结果的：

````python
# 例题：100 > 1 and -10 ** 2 <= -100 or "foo" == "foo" + "bar"
# 我们来一步步拆解

print("表达式: 100 > 1 and -10 ** 2 <= -100 or 'foo' == 'foo' + 'bar'")
print()

# Step 1: ** 优先级最高
step1 = -10 ** 2
print(f"Step 1: -10 ** 2 = {step1}   （** 优先于负号，等价于 -(10**2)）")

# Step 2: 字符串拼接
step2 = "foo" + "bar"
print(f"Step 2: 'foo' + 'bar' = '{step2}'")

# Step 3: 比较运算符
print(f"Step 3: 100 > 1 = {100 > 1}")
print(f"       -100 <= -100 = {-100 <= -100}")
print(f"       'foo' == 'foobar' = {'foo' == 'foobar'}")

# Step 4: and 和 or
print(f"Step 4: True and True or False")
print(f"       = (True and True) or False  (and 优先于 or)")
print(f"       = True or False")
print(f"       = {True and True or False}")

# 验证
result = 100 > 1 and -10 ** 2 <= -100 or "foo" == "foo" + "bar"
print(f"\n最终结果: {result}")
````

### ⚡ 短路求值（Short-Circuit Evaluation）

Python 在计算 `and` 和 `or` 时有一个聪明的优化——**短路求值**：

- `True or ???` → 不管 `???` 是什么，结果一定是 `True`，所以 `???` 根本**不执行**
- `False and ???` → 不管 `???` 是什么，结果一定是 `False`，所以 `???` 根本**不执行**

这就像你判断"门开着或者…"——只要门开着，后面都不用看了。

````python
# 短路求值演示

# 例1: True or ... → 短路，右边不执行
print("=== 短路示例 1 ===")
result = True or print("这行会不会打印？")
print(f"True or (print) = {result}")
print("→ 右边没执行，因为 True or 任何东西 = True")

# 例2: False and ... → 短路，右边不执行
print("\n=== 短路示例 2 ===")
result = False and print("这行会不会打印？")
print(f"False and (print) = {result}")
print("→ 右边没执行，因为 False and 任何东西 = False")

# 例3: 短路求值的实用场景
print("\n=== 实用场景 ===")
name = ""  # 空字符串
# 如果 name 非空，才取第一个字符；否则用默认值
first_char = name and name[0] or "无名"
print(f"name='{name}' → first_char='{first_char}'")

name = "Alice"
first_char = name and name[0] or "无名"
print(f"name='{name}' → first_char='{first_char}'")
````

## 第二部分：决策结构

到目前为止，我们的程序都是"一条路走到底"——从上往下依次执行每一行。

但现实中我们需要**做选择**：
- 如果下雨 → 带伞
- 如果成绩 >= 60 → 及格
- 如果今天是周末 → 睡懒觉，否则 → 早起

这就是**决策结构（Decision Structure）**——让程序根据条件走不同的"路"。

---

### 🌲 `if` 语句 — 最简单的分支

**语法**：
```python
if <条件表达式>:
    # 条件为 True 时执行的代码
    # 注意：前面必须缩进（4个空格或1个Tab）
```

**流程图**：
```
      ┌─────────┐
      │  条件?  │
      └────┬────┘
      True │ False
      ┌────┘
      ↓
  ┌───────┐
  │ 代码块 │
  └───┬───┘
      ↓
  (继续执行)
```

当条件为 `True` 时，执行缩进的代码块；为 `False` 时，跳过。

````python
# if 语句示例：判断正数
number = int(input("请输入一个数字："))

if number >= 0:
    print(f"{number} 是正数（或零）。")

print("程序结束。")  # 这行不管条件是否成立都会执行
````

### 🌲🌲 `if-else` 语句 — 二选一

**语法**：
```python
if <条件表达式>:
    # 条件为 True 时执行
else:
    # 条件为 False 时执行
```

**流程图**：
```
      ┌─────────┐
      │  条件?  │
      └────┬────┘
      True │ False
      ┌────┘────┐
      ↓         ↓
  ┌───────┐ ┌───────┐
  │ 代码A  │ │ 代码B  │
  └───┬───┘ └───┬───┘
      └────┬────┘
           ↓
      (继续执行)
```

````python
# if-else 示例：判断正负数
number = int(input("请输入一个数字："))

if number >= 0:
    print(f"{number} 是正数（或零）。")
else:
    print(f"{number} 是负数。")

print("程序结束。")
````

### 🎬 经典示例：红蓝药丸

> 出自《黑客帝国》：吃蓝色药丸，故事结束；吃红色药丸，留在仙境。

````python
# 红蓝药丸 — The Matrix
pill = input("选一颗药丸（blue / red）：")

if pill == "blue":
    print("💊 故事结束。你醒来后什么都不记得。")
else:
    print("💊 你留在仙境。我将带你看看兔子洞有多深……")
````

### 🔢 练习：两数比较

输入两个数，如果相等打印和，如果不相等打印积。

````python
# 练习：输入两个数
a = float(input("请输入第一个数："))
b = float(input("请输入第二个数："))

if a == b:
    print(f"两数相等，它们的和是 {a + b}")
else:
    print(f"两数不等，它们的积是 {a * b}")
````

### 🌲🌲🌲 `if-elif-else` 语句 — 多选一

当有三个或更多分支时，用 `elif`（else if 的缩写）：

**语法**：
```python
if 条件1:
    # 条件1成立时执行
elif 条件2:
    # 条件1不成立但条件2成立时执行
elif 条件3:
    # 可以有很多个 elif
else:
    # 以上条件都不成立时执行
```

> ⚠️ 重要：`elif` 是**按顺序检查**的，第一个为 `True` 的条件被执行后，后面的条件**不再检查**。所以 **顺序很重要**！

````python
# if-elif-else 示例：判断数字的正负零
number = int(input("请输入一个数字："))

if number > 0:
    print(f"{number} 是正数。")
elif number < 0:
    print(f"{number} 是负数。")
else:  # number == 0
    print(f"{number} 是零。")

print("程序结束。")
````

### 🍂 经典示例：四季判断

输入季节名称，输出对应的风景：

````python
# 四季判断
season = input("输入一个季节（Summer/Winter/Fall/Spring）：")

if season == "Summer":
    print("☀️ Sun — 阳光明媚！")
elif season == "Winter":
    print("❄️ Snow — 大雪纷飞！")
elif season == "Fall":
    print("🍂 Leaves — 落叶缤纷！")
else:  # Spring
    print("🌸 Flowers — 春暖花开！")
````

### ⚠️ `elif` 链 vs 多个 `if`

这是一个容易出错的点：`elif` 链只执行**第一个为 True 的分支**，而多个独立的 `if` 会**逐一检查**。

````python
# 对比：elif 链 vs 多个 if
value = 100

print("=== 使用 elif 链 ===")
if value > 10:
    print("大于 10")
elif value > 50:
    print("大于 50")   # 这个条件虽然也是 True，但不会执行
elif value > 80:
    print("大于 80")   # 也不会执行

print("\n=== 使用多个独立 if ===")
if value > 10:
    print("大于 10")
if value > 50:
    print("大于 50")   # 会执行
if value > 80:
    print("大于 80")   # 会执行
````

### 🪆 嵌套 `if` 语句 — 条件里面套条件

在一个 `if` 的代码块里面再写 `if`，就像俄罗斯套娃。

**经典例子：颜色混合器**

输入两种颜色，输出混合后的颜色：

|  | Red | Blue | Yellow |
|:---|:---:|:---:|:---:|
| **Red** | Red | Purple | Orange |
| **Blue** | Purple | Blue | Green |
| **Yellow** | Orange | Green | Yellow |

````python
# 颜色混合器 — 嵌套 if 语句
color1 = input("输入第一种颜色（Red/Blue/Yellow）：")
color2 = input("输入第二种颜色（Red/Blue/Yellow）：")

# 如果两个颜色相同
if color1 == color2:
    print(f"混合色是 {color1}")
elif color1 == "Red":
    if color2 == "Blue":
        print("混合色是 Purple（紫色）🟣")
    else:  # color2 是 Yellow
        print("混合色是 Orange（橙色）🟠")
elif color1 == "Blue":
    if color2 == "Red":
        print("混合色是 Purple（紫色）🟣")
    else:  # color2 是 Yellow
        print("混合色是 Green（绿色）🟢")
elif color1 == "Yellow":
    if color2 == "Red":
        print("混合色是 Orange（橙色）🟠")
    else:  # color2 是 Blue
        print("混合色是 Green（绿色）🟢")
else:
    print("请输入 Red、Blue 或 Yellow 之一！")
````

````python
# ====== 练习 3：三个数比大小 ======
n1 = float(input("请输入第一个数字："))
n2 = float(input("请输入第二个数字："))
n3 = float(input("请输入第三个数字："))

if n1 >= n2 and n1 >= n3:
    largest = n1
elif n2 >= n1 and n2 >= n3:
    largest = n2
else:
    largest = n3

print(f"最大的数是 {largest}")

# 更简单的方法：
print(f"用 max() 验证：{max(n1, n2, n3)}")
````

````python
# ====== 练习 4：BMI 计算器 ======
height = float(input("请输入身高（米）："))
weight = float(input("请输入体重（公斤）："))

bmi = weight / (height ** 2)
print(f"\n你的 BMI 是 {bmi:.1f}")

if bmi < 18.5:
    print("体型：偏瘦 🏃")
elif bmi < 24:
    print("体型：正常 ✅")
elif bmi < 28:
    print("体型：偏胖 ⚠️")
else:
    print("体型：肥胖 ❗")
````

````python
# ====== 练习 5（挑战）：简易计算器 ======
num1 = float(input("请输入第一个数字："))
op = input("请输入运算符（+ - * /）：")
num2 = float(input("请输入第二个数字："))

if op == "+":
    result = num1 + num2
elif op == "-":
    result = num1 - num2
elif op == "*":
    result = num1 * num2
elif op == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "错误：除数不能为零！"
else:
    result = "错误：不支持的运算符！"

print(f"\n{num1} {op} {num2} = {result}")
````

## 📚 本课总结

| 知识点 | 关键内容 |
|:---|:---|
| 🔘 布尔值 | `True` / `False`，是数字也是逻辑 |
| ⚖️ 比较运算符 | `==` `!=` `>` `<` `>=` `<=` |
| 🔍 成员运算符 | `in` / `not in` |
| 🧠 逻辑运算符 | `not`（取反）、`and`（全真才真）、`or`（有真就真） |
| 📐 优先级 | `()` → `**` → `* /` → `+ -` → 比较 → `not` → `and` → `or` |
| ⚡ 短路求值 | `True or x` 不执行 `x`；`False and x` 不执行 `x` |
| 🌲 `if` | 条件为 `True` 时执行代码块 |
| 🌲🌲 `if-else` | 两条路二选一 |
| 🌲🌲🌲 `if-elif-else` | 多条路选一条，顺序很重要 |
| 🪆 嵌套 `if` | 条件里面套条件 |

> 🎯 **核心思想**：决策结构让程序不再是"一条路走到黑"，而是能根据不同情况做出不同反应——这就是程序的"智能"的起点！
