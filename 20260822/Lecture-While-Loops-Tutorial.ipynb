## 三种控制结构

程序有三种基本控制结构：

```
┌──────────┐    ┌──────────────┐    ┌──────────────┐
│  顺序     │    │   决策       │    │   循环       │
│ Sequence  │    │  Decision    │    │  Iteration   │
├──────────┤    ├──────────────┤    ├──────────────┤
│ 语句1    │    │    条件?     │    │    条件?     │
│   ↓      │    │   ↙    ↘    │    │   ↙    ↘    │
│ 语句2    │    │ True  False  │    │ True  False  │
│   ↓      │    │  ↓     ↓    │    │  ↓     ↓    │
│ 语句3    │    │语句A  语句B  │    │循环体  退出  │
└──────────┘    └──────────────┘    │  ↓         │
                                    │ 回到条件───┘
                                    └──────────────┘
```

上次学了**决策**（if/else），今天学**循环**（while）——让程序拥有"重复做"的能力！

## 第一部分：为什么需要循环？

### 问题：打印 1 到 9

如果不用循环，我们只能这样写：

```python
n = 1
print(n)   # 1
n = n + 1
print(n)   # 2
n = n + 1
print(n)   # 3
# ... 重复 9 次
```

**问题来了**：如果要打印 1 到 1,000,000 呢？你要复制粘贴 100 万行吗？

这就是**循环（Loop）** 的用武之地——"一段代码，重复执行，直到条件不满足"。

### 关键概念：Iteration vs Loop

| 术语 | 含义 |
|:---|:---|
| **Iteration（迭代）** | 重复执行一组语句的**行为** |
| **Loop（循环）** | 实现迭代的**代码结构** |

一个 Loop 可能包含很多次 Iteration——就像一首歌（Loop）反复播了 100 遍（100 次 Iteration）。

## 第二部分：while 循环基础

### 2.1 while 循环的工作原理

```
      ┌─────────┐
      │  条件?  │
      └────┬────┘
      True │ False
      ┌────┘
      ↓         → 退出循环
  ┌───────┐
  │ 循环体 │
  └───┬───┘
      │
      └──→ 回到条件判断
```

**核心规则**：
- 条件为 `True` → 执行循环体 → 回到条件
- 条件为 `False` → 退出循环
- ⚠️ 如果条件**永远为 True** → 死循环！按 `Ctrl+C` 强制停止

### 2.2 语法

```python
while <条件表达式>:
    # 条件为 True 时重复执行
    # 注意缩进！

# 缩进恢复后，循环结束
```

````python
# ====== 第一个 while 循环：计数器 ======
count = 1                    # ① 起始值（哨兵变量）

while count <= 5:            # ② 条件：count <= 5 时继续
    print("Count is:", count) # ③ 循环体：打印当前值
    count = count + 1         # ④ 更新：每次加 1（非常重要！）

print("循环结束！")
````

### 2.3 while 循环的三个要素

每个 while 循环都需要三个部分：

```python
count = 1              # ① 初始化（起始值）
while count <= 5:      # ② 条件（什么时候继续）
    print(count)
    count = count + 1  # ③ 更新（向结束靠近）
```

| 要素 | 作用 | 忘了会怎样 |
|:---|:---|:---|
| ① 初始化 | 设起始值 | 变量未定义 → NameError |
| ② 条件 | 控制何时停止 | — |
| ③ 更新 | 向终止条件靠近 | **死循环！** |

> 🚨 忘记第③步是最常见的错误！`count` 永远不变，条件永远为 True，程序永远停不下来。

````python
# ⚠️ 死循环示例 — 不要运行！
# count = 1
# while count <= 5:
#     print(count)
#     # 忘了写 count = count + 1
#     # → count 永远是 1，条件永远 True，程序卡死！

print("上面的代码被注释掉了，因为会导致死循环！")
print("如果不小心写了死循环，按 Ctrl+C 强制停止。")
````

count = 1

## 第三部分：while 循环实战

### 3.1 密码验证 — 结合 input()

````python
# ====== 密码验证循环 ======
password = "ICPFALL25"      # 正确的密码
stop = False                 # 哨兵：控制是否退出循环
user_input = ""              # 用户输入

while not stop:              # not stop = not False = True（继续循环）
    user_input = input("请输入密码：")
    if user_input == password:
        stop = True          # 密码正确 → 改变哨兵 → 下次循环退出
        print("✅ 访问授权！")
    else:
        print("❌ 密码错误，请重试。")
````

### 3.2 密码验证 — 限制尝试次数

````python
# ====== 密码验证（最多 3 次尝试）======
password = "ICPFALL25"
attempts = 0                 # 尝试次数计数器
max_attempts = 3             # 最大允许次数
access_granted = False       # 是否已授权

while not access_granted and attempts < max_attempts:
    user_input = input(f"请输入密码（剩余 {max_attempts - attempts} 次）：")
    attempts = attempts + 1   # 尝试次数 +1
    
    if user_input == password:
        access_granted = True
        print("✅ 访问授权！")
    elif attempts < max_attempts:
        print(f"❌ 密码错误，还有 {max_attempts - attempts} 次机会。")

if not access_granted:
    print("🚫 尝试次数已用完，访问被拒绝。")
````

### 3.3 倒计时

````python
# ====== 倒计时程序 ======
a = int(input("请输入一个正整数："))

while a >= 0:                # 只要 a >= 0 就继续
    print(a)
    a = a - 1                 # 每次减 1（向 0 靠近）

print("🚀 发射！")
````

### 3.4 按步长计数：从 2 数到 8，每次加 2

````python
# ====== 方法 1：直接改步长 ======
count = 2
while count <= 8:
    print(count)
    count = count + 2         # 每次加 2
print("完成")

print()

# ====== 方法 2：用 if 过滤偶数 ======
count = 2
while count <= 8:
    if count % 2 == 0:        # 如果是偶数才打印
        print(count)
    count = count + 1
print("完成")
````

### 3.5 十进制转二进制

````python
# ====== 十进制 → 二进制转换器 ======
stop = False

while not stop:
    a = input("请输入一个整数（或输入 'stop' 退出）：")
    
    if a != 'stop':
        # bin() 是 Python 内置函数，返回二进制字符串（如 '0b1010'）
        print(f"{a} 的二进制是：{bin(int(a))}")
    else:
        stop = True

print("再见！")
````

## 第四部分：累加器（Accumulator）

### 4.1 什么是累加器？

**累加器**是一个在循环中不断累积结果的变量——像一个"存钱罐"，每次循环往里面丢一点。

```
total = 0          ← 空罐子
循环第 1 次: total = total + 1  → total = 1
循环第 2 次: total = total + 2  → total = 3
循环第 3 次: total = total + 3  → total = 6
循环第 4 次: total = total + 4  → total = 10
```

````python
# ====== 累加器示例：求 0+1+2+3+4 的和 ======
num = 0          # 哨兵变量（控制循环）
total = 0        # 累加器（累积结果）

while num < 5:
    total = total + num   # 把当前 num 加入总和
    num = num + 1         # num 前进

print(f"0+1+2+3+4 = {total}")
````

### 4.2 哨兵 vs 累加器

| 角色 | 作用 | 变化方式 |
|:---|:---|:---|
| **哨兵 (Sentinel)** | 控制循环何时停止 | 每次加固定值（如 `+1`）|
| **累加器 (Accumulator)** | 累积计算结果 | 每次加上不同的值 |

````python
# ====== 课堂练习 5.4：分析这段代码 ======
num = 1
total = 1

while num < 5:
    total = total * num    # 累乘！
    num = num + 1

print(f"结果：{total}")
print()
print("分析：")
print("  哨兵变量：num（从 1 到 4，控制循环）")
print("  累加器变量：total（累积乘积）")
print("  计算过程：total = 1*1*2*3*4 = 24")
print("  这其实就是 4! (4 的阶乘)！")
````

## 第五部分：增强赋值运算符

### 5.1 简写语法

每次都写 `x = x + 1` 太啰嗦了。Python 提供了简写：

| 运算符 | 等价于 | 示例 | 结果 |
|:---:|:---|:---|:---|
| `+=` | `x = x + 值` | `x += 5` | x 增加 5 |
| `-=` | `x = x - 值` | `y -= 2` | y 减少 2 |
| `*=` | `x = x * 值` | `z *= 10` | z 乘 10 |
| `/=` | `x = x / 值` | `a /= b` | a 除以 b |
| `%=` | `x = x % 值` | `c %= 3` | c 对 3 取余 |

````python
# ====== 增强赋值运算符演示 ======

# += 加等
x = 10
x += 5          # 等价于 x = x + 5
print(f"x += 5  →  x = {x}")

# -= 减等
y = 20
y -= 3          # 等价于 y = y - 3
print(f"y -= 3  →  y = {y}")

# *= 乘等
z = 7
z *= 10         # 等价于 z = z * 10
print(f"z *= 10 →  z = {z}")

# /= 除等
a = 100
a /= 4          # 等价于 a = a / 4
print(f"a /= 4  →  a = {a}")

# %= 取余等
c = 17
c %= 3          # 等价于 c = c % 3
print(f"c %= 3  →  c = {c}")
````

````python
# ====== 用增强赋值改写累加器 ======
num = 0
total = 0

while num < 5:
    total += num      # 比 total = total + num 更简洁
    num += 1          # 比 num = num + 1 更简洁

print(f"0+1+2+3+4 = {total}")

# 用增强赋值改写倒计时
a = int(input("\n请输入倒计时起始数："))
while a >= 0:
    print(a, end=" ")
    a -= 1           # 比 a = a - 1 更简洁
print("\n🚀 发射！")
````

## 第六部分：跳过特定条件的数字

### 问题：打印 1 到最大值，但跳过包含数字 4 的数

例如 max=20，跳过 4、14（包含 4），打印 1,2,3,5,6,...,13,15,...

````python
# ====== 跳过包含数字 4 的数 ======
max_val = int(input("请输入最大值："))
num = 1

while num <= max_val:
    # 把数字转成字符串，检查是否包含字符 '4'
    if '4' not in str(num):
        print(num, end=" ")
    num += 1

print("\n完成！")
````

### 7.1 什么是模块？

Python 内置了一些函数（`print`、`input`、`int`…），但更多功能藏在**模块（Module）** 里。模块就像工具箱——你需要的时候拿出来用。

```python

```

Python 有超过 200 个标准模块！今天学 4 个最常用的。

### 7.2 math 模块 — 数学工具箱

## 🎮 综合实战：猜数字游戏（完整版）

综合运用 while 循环 + 哨兵 + 累加器 + random 模块：

````python
# ====== 猜数字游戏（完整版）======
import random

print("=" * 40)
print("    🎮 猜数字游戏")
print("    我想了一个 1~100 之间的数字")
print("=" * 40)

secret = random.randint(1, 100)   # 随机生成目标数字
guess = 0                          # 用户猜的数
attempts = 0                       # 累加器：猜了几次

while guess != secret:
    guess = int(input("你猜是多少？"))
    attempts += 1                  # 猜测次数 +1
    
    if guess < secret:
        print("  📈 太小了！往上猜")
    elif guess > secret:
        print("  📉 太大了！往下猜")

print(f"\n🎉 恭喜！答案就是 {secret}！")
print(f"   你一共猜了 {attempts} 次")

# 评价
if attempts <= 5:
    print("   ⭐ 太厉害了！")
elif attempts <= 10:
    print("   👍 不错！")
else:
    print("   🤔 下次加油！")
````

````python
# ====== 练习 5（挑战）：简易 ATM ======
balance = 1000.0    # 初始余额

print("=" * 30)
print("    🏦 简易 ATM")
print("=" * 30)

choice = 0
while choice != 4:
    print("\n1. 查询余额")
    print("2. 存款")
    print("3. 取款")
    print("4. 退出")
    choice = int(input("请选择操作："))
    
    if choice == 1:
        print(f"💰 当前余额：{balance:.2f} 元")
    elif choice == 2:
        amount = float(input("存款金额："))
        if amount > 0:
            balance += amount
            print(f"✅ 存款成功！当前余额：{balance:.2f} 元")
        else:
            print("❌ 金额必须大于 0！")
    elif choice == 3:
        amount = float(input("取款金额："))
        if amount > balance:
            print("❌ 余额不足！")
        elif amount <= 0:
            print("❌ 金额必须大于 0！")
        else:
            balance -= amount
            print(f"✅ 取款成功！当前余额：{balance:.2f} 元")
    elif choice == 4:
        print("👋 感谢使用，再见！")
    else:
        print("❌ 无效选择，请重新输入！")
````
