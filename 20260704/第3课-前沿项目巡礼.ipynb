# 🖥️ 项目一：Claude Code

> [github.com/anthropics/claude-code](https://github.com/anthropics/claude-code)

## 它是什么？

一个**住在终端里的 AI 编程助手**。你跟它用中文聊天，它帮你写代码、修 bug、提交 Git。

```
你：帮我写一个计算器
Claude Code：写好了，运行 calculator.py 试试
你：不对，少了个除法
Claude Code：已修复，加了 / 运算
```

## 为什么这么火？

| 数据 | 含义 |
|:---|:---|
| ⭐ 136,000 | GitHub 全站最热项目之一 |
| 🍴 21,800 | 2万多人复制了它 |
| 🐍 80% Python | 核心是 Python 写的 |

## 🤯 跟 Copilot 有什么区别？

| | Copilot | Claude Code |
|:---|:---|:---|
| 在哪用 | VSCode 里 | **终端里**（更底层） |
| 能做什么 | 补全代码、聊天 | 读写文件、执行命令、Git 操作 |
| 自主性 | 你告诉它每步做什么 | 它能**自己规划并执行多步任务** |

> 简单说：Copilot 是你的副驾驶，Claude Code 是**能自己开车的 AI**。

# 🤖 项目二：CUA（Computer-Use Agent）

> [github.com/trycua/cua](https://github.com/trycua/cua)

## 它是什么？

**让 AI 能操控你的电脑桌面。** 点击、打字、打开软件——AI 全自动完成。

```
你对 Claude Code 说：
"帮我打开浏览器，登录 Gmail，看看最近3封未读邮件"

Claude Code + CUA：
1. 自动打开 Chrome
2. 导航到 gmail.com
3. 输入账号密码登录
4. 读取未读邮件标题
5. 回复你："最近3封是：xxx"
```

## 怎么做到的？

CUA 提供了 **MCP 服务器**（Model Context Protocol），接入 Claude Code 后，AI 就获得了"手"和"眼睛"：

```
Claude Code  ←→  CUA MCP Server  ←→  你的桌面
  (大脑)         (翻译官)            (手和眼睛)
```

# 🔌 项目三：Claude Code Proxy

> [github.com/Nolan180940/Claude-Code-Proxy-SiliconFlowCN](https://github.com/Nolan180940/Claude-Code-Proxy-SiliconFlowCN)

## 它是什么？

你老师的开源项目！**让 Claude Code 用上国产大模型**（DeepSeek、GLM），而不是只能付费买 Claude。

## 为什么需要它？

```
原本：
Claude Code ──→ Claude API（要付费、要翻墙）

有了 Proxy：
Claude Code ──→ Proxy ──→ DeepSeek / GLM（国内、便宜、快）
```

## 工作原理

```
你说的话 ──→ Claude Code ──→ Proxy 翻译 ──→ DeepSeek 处理
                                                      ↓
屏幕上的结果 ←── Claude Code ←── Proxy 翻译回来 ←── 返回结果
```

就像一个**同声传译**，让说英语的 Claude Code 能和说中文协议的 DeepSeek 沟通。

## 技术栈

| 技术 | 作用 |
|:---|:---|
| Node.js | 代理服务器 |
| SiliconFlow | 国产模型 API 平台 |
| DeepSeek V4 Pro | 主力模型 |
| GLM | 备用模型 |
