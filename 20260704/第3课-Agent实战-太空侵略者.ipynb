## 🎮 游戏描述

这是一个经典的太空射击游戏。玩家控制一艘飞船在屏幕底部，屏幕顶部有 5×5 排列的外星人。外星人会左右移动并向玩家射击。玩家必须消灭所有外星人才能获胜。

## 🖼️ 素材清单

| 文件 | 用途 |
|:---|:---|
| `bg.png` | 星空背景（600×800） |
| `spaceship.png` | 玩家飞船 |
| `alien1.png` ~ `alien5.png` | 5 种外星人外观（随机使用） |
| `bullet.png` | 玩家子弹 |
| `alien_bullet.png` | 外星人子弹 |
| `exp1.png` ~ `exp5.png` | 爆炸动画 5 帧序列 |
| `laser.wav` | 玩家发射音效 |
| `explosion.wav` | 外星人被击中音效 |
| `explosion2.wav` | 飞船被击中音效 |

## 🕹️ 玩法

- 窗口 600×800，标题 "Space Invaders"，60 FPS
- 玩家用**左右方向键**移动飞船，**空格键**发射子弹
- 飞船不能移出屏幕边界
- 飞船有 3 条命，屏幕下方显示血条（绿色血量 + 红色底色）
- 外星人 5 行 × 5 列排列在屏幕上方，自动左右移动，到达边缘时换向
- 外星人随机向玩家射击
- 子弹击中目标时播放爆炸动画（5 帧序列，播完消失）
- 击中音效：玩家子弹击中 → `explosion.wav`，外星人子弹击中飞船 → `explosion2.wav`，发射子弹 → `laser.wav`
- 音效音量统一设为 0.25

## 🏆 胜负条件

- 消灭全部外星人 → 显示 "YOU WIN!"
- 飞船血量归零（被击中 3 次）→ 飞船大爆炸，显示 "GAME OVER!"
- 开局前 3 秒倒计时，显示 "GET READY!" 和倒计时数字

## 🎨 视觉细节

- 爆炸有三种大小，根据触发原因自动选择合适的尺寸
- 5 种外星人外观随机混搭，让画面更丰富
- 字体用 Constantia，大小 30 和 40，白色文字
