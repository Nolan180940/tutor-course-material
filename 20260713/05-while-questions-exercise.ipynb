## 第 1 题：预测输出

### 程序 1 — 字符串格式化

```python
# ====== 第 1 题 — 程序 1：预测输出 ======
# 题目：预测以下程序的输出结果。

# 参考答案：
a = "python"
b = ""
i = len(a)                         # i = 6

while i > 0:
    # 每次循环：取 a 的前 i 个字符，右对齐，宽度为 len(a)
    # 用 '-' 填充空白
    b = b + str(i) + ":" + '{:->{w}}'.format(a[:i], w=len(a)) + "\n"
    i -= 1

print(b + "0:" + "-" * len(a))

print("\n解析：")
print("  i=6: '6:-python'")
print("  i=5: '5:-pytho'")
print("  i=4: '4:-pyth'")
print("  i=3: '3:-pyt'")
print("  i=2: '2:-py'")
print("  i=1: '1:-p'")
print("  i=0: '0:------'")
```

### 程序 2 — 嵌套循环打印星号

```python
# ====== 第 1 题 — 程序 2：预测输出 ======
# 题目：预测嵌套 while 循环打印的星号图案。

# 参考答案：
m = 6
n = 3
i = 1

while i < m:                       # i: 1→5
    j = 1
    while j <= i and i <= n:       # i≤3 时：每行 i 个星号
        print('*', end='|')
        j += 1
    j += 1
    while j <= m and i > n:        # i>3 时：每行 (m-j+1) 个星号
        print('*', end='|')
        j += 1
    print()
    i += 1

print("\n解析：")
print("  i=1 (≤3): 打印 1 个 *")
print("  i=2 (≤3): 打印 2 个 *")
print("  i=3 (≤3): 打印 3 个 *")
print("  i=4 (>3): 打印 2 个 *")
print("  i=5 (>3): 打印 1 个 *")
```

### 程序 3 — 类斐波那契序列

```python
# ====== 第 1 题 — 程序 3：预测输出 ======
# 题目：预测以下程序的输出。

# 参考答案：
a = 0
b = 1

while a < 10:
    print(a)                       # 打印当前 a
    c, a = a, b                     # c=旧a, a=b
    b = c + b                       # b = 旧a + b

print("\n解析：这是斐波那契数列！")
print("  初始: a=0, b=1")
print("  第1轮: print(0), c=0,a=1, b=0+1=1")
print("  第2轮: print(1), c=1,a=1, b=1+1=2")
print("  第3轮: print(1), c=1,a=2, b=1+2=3")
print("  第4轮: print(2), c=2,a=3, b=2+3=5")
print("  第5轮: print(3), c=3,a=5, b=3+5=8")
print("  第6轮: print(5), c=5,a=8, b=5+8=13")
print("  第7轮: print(8), ... (8 < 10 继续)")
print("  输出: 0 1 1 2 3 5 8")
```

## 第 2 题：字符图案

阅读以下程序，预测输出结果。提示：用网格画出每个字符的位置。

```python
# ====== 第 2 题：预测字符图案输出 ======
# 题目：预测嵌套 while 循环 + chr()/ord() 的输出。

# 参考答案：
m = 5
n = 3
x = ord('A')                       # x = 65 ('A' 的 ASCII 码)
s = ""
i = 1

while i < m:                       # i: 1→4
    j = 1
    while j <= i and i <= n:       # i≤3 时：每行 i 个字母
        s += chr(x) + '|'
        j += 1
        x += 1
    j += i
    while j <= m and not i > n:    # 等价于 j≤m and i≤n
        s += chr(x) + '|'
        j += 1
        x += 1
    print(s)
    i += 1

print("\n解析（逐字符追踪）：")
print("  每个字母按字母表顺序递增 (A→B→C→...)")
print("  i=1 (≤3): 1个字母 + (5-1-1)=3个 = 4个字母")
print("  i=2 (≤3): 2个字母 + (5-2-2)=1个 = 3个字母")
print("  i=3 (≤3): 3个字母 + (5-3-3)=-1(不执行) = 3个字母")
print("  i=4 (>3): j≤i 时 4≤4? → j=4 不满足 j≤i? 退出")
```

## 第 3 题：有趣的数字

### 3.1 自幂数（Narcissistic Number）

**定义**：一个 n 位数，等于其各位数字的 n 次幂之和。

例如：
- $371 = 3^3 + 7^3 + 1^3 = 27 + 343 + 1 = 371$（3 位数）
- $9474 = 9^4 + 4^4 + 7^4 + 4^4 = 6561 + 256 + 2401 + 256 = 9474$（4 位数）

编写程序，输入一个正整数，判断它是否为自幂数。

```python
# ====== 第 3.1 题：自幂数判断 ======
# 题目：输入一个正整数，判断是否为自幂数（Narcissistic Number）。

# 参考答案：
num = input("请输入一个正整数：")
n = len(num)                       # 数字的位数
num_int = int(num)

# 计算各位数字的 n 次幂之和
total = 0
i = 0
while i < n:
    digit = int(num[i])            # 取出第 i 位数字
    total += digit ** n            # 累加 digit 的 n 次方
    i += 1

# 判断
if total == num_int:
    print(f"{num_int} 是自幂数 ✅")
else:
    print(f"{num_int} 不是自幂数 ❌（各位{n}次幂之和 = {total}）")

# 验证几个已知的自幂数
print("\n已知自幂数验证：")
for n in [153, 370, 371, 407, 1634, 8208, 9474]:
    digits = [int(d) for d in str(n)]
    power_sum = sum(d ** len(str(n)) for d in digits)
    status = "✅" if power_sum == n else "❌"
    print(f"  {n}: {status}")
```

### 3.2 Kaprekar 数

**定义**：对于一个 d 位正整数 n，将 $n^2$ 拆成左右两部分（右边 d 位），两部分之和等于 n。

例如：
- $9^2 = 81$，$8 + 1 = 9$ ✅
- $45^2 = 2025$，$20 + 25 = 45$ ✅
- $297^2 = 88209$，$88 + 209 = 297$ ✅

```python
# ====== 第 3.2 题：Kaprekar 数判断 ======
# 题目：输入一个正整数，判断是否为 Kaprekar 数。

# 参考答案：
n = int(input("请输入一个正整数："))
d = len(str(n))                    # n 的位数
square = n ** 2                    # n 的平方
square_str = str(square)

# 把平方数拆成左右两部分：右边 d 位，左边剩余
right_len = d
left_len = len(square_str) - d

if left_len <= 0:
    print(f"{n} 不是 Kaprekar 数 ❌（平方数位数不足）")
else:
    right_part = int(square_str[left_len:])   # 右边 d 位
    left_part = int(square_str[:left_len])     # 左边剩余
    
    if left_part + right_part == n:
        print(f"{n} 是 Kaprekar 数 ✅")
        print(f"  {n}² = {square}, {left_part} + {right_part} = {n}")
    else:
        print(f"{n} 不是 Kaprekar 数 ❌")
        print(f"  {n}² = {square}, {left_part} + {right_part} = {left_part + right_part} ≠ {n}")

# 验证已知 Kaprekar 数
print("\n已知 Kaprekar 数验证（前几个）：")
test_numbers = [9, 45, 55, 99, 297, 703, 999, 2223, 2728, 4879]
for num in test_numbers:
    s = str(num ** 2)
    d = len(str(num))
    if len(s) > d:
        left = int(s[:-d]) if s[:-d] else 0
        right = int(s[-d:])
        if left + right == num:
            print(f"  {num}²={num**2} → {left}+{right}={num} ✅")
```

## 第 4 题：字符串处理

### 4.1 预测程序功能与输出

阅读以下程序，用一句话概括其功能，并预测给定输入下的输出结果。

```python
# ====== 第 4.1 题：分析程序功能并预测输出 ======
# 题目：分析以下程序的功能，预测各输入的输出。

# 参考答案：
# 功能：去除字符串中连续重复的字符（相邻相同字符只保留一个）

def deduplicate(N):
    """去除连续重复字符"""
    i, j = 0, 0
    result = ""
    while i < len(N):
        j = i + 1
        while j < len(N) and N[j] == N[i]:   # 跳过所有与 N[i] 相同的连续字符
            j = j + 1
        result += N[i]
        i = j                                  # 跳到下一个不同字符
    return result

# 测试
test_inputs = [
    "abcdeabcdeabcde",
    "aabbccddeeff",
    "abcaabbccaaabbbcccddd"
]

print("程序功能：去除字符串中连续重复的字符（相邻相同只保留一个）\n")
print("输入                    →  输出")
print("-" * 50)
for inp in test_inputs:
    out = deduplicate(inp)
    print(f"{inp:25s} →  {out}")
```

### 4.2 统计子串出现次数

编写程序：输入两个字符串，统计第二个字符串在第一个字符串中出现的次数。

```python
# ====== 第 4.2 题：统计子串出现次数 ======
# 题目：输入两个字符串，统计第二个在第一个中出现的次数。

# 参考答案：
main_str = input("请输入一个字符串：")
sub_str = input("请输入另一个字符串：")

count = 0
i = 0

# 方法：逐位滑动窗口检查
while i <= len(main_str) - len(sub_str):
    # 检查从 i 开始的子串是否匹配
    if main_str[i:i + len(sub_str)] == sub_str:
        count += 1
    i += 1

print(f"字符串 '{sub_str}' 出现了 {count} 次。")

# 示例验证
print("\n示例验证：")
print("  主串: abcdeabcdeabcde, 子串: abc →", end=" ")
s = "abcdeabcdeabcde"
sub = "abc"
c = 0
i = 0
while i <= len(s) - len(sub):
    if s[i:i+len(sub)] == sub:
        c += 1
    i += 1
print(c, "次")
```

## 第 5 题：ISBN 校验

ISBN（国际标准书号）的最后一位是**校验位（check digit）**，用来检测 ISBN 输入错误。

### ISBN-10（10 位）校验规则

1. 计算前 9 位的加权和 s，权重从 10 递减到 2
2. 计算 n = (11 - s % 11) % 11，使 s + n 能被 11 整除
3. 如果 n = 10，校验位为字母 `'X'`；否则校验位 = n

### ISBN-13（13 位）校验规则

1. 计算前 12 位的加权和 s，奇数位权重 1，偶数位权重 3
2. 计算 n = (10 - s % 10) % 10，使 s + n 能被 10 整除
3. 校验位 = n

```python
# ====== 第 5 题：ISBN 校验 ======
# 题目：输入一个 ISBN 号码，判断其是否为合法的 ISBN-10 或 ISBN-13。

# 参考答案：
isbn = input("请输入一个 ISBN 号码：").strip()

# 检查长度
if len(isbn) == 10:
    # ── ISBN-10 校验 ──
    # 计算前 9 位的加权和（权重 10→2）
    s = 0
    i = 0
    while i < 9:
        digit = int(isbn[i])
        s += digit * (10 - i)
        i += 1
    
    # 期望的校验位
    n = (11 - s % 11) % 11
    expected_check = 'X' if n == 10 else str(n)
    actual_check = isbn[9]
    
    if actual_check == expected_check:
        print("✅ 合法的 ISBN-10")
    else:
        print(f"❌ 非法的 ISBN-10（期望校验位：'{expected_check}'，实际：'{actual_check}'）")

elif len(isbn) == 13:
    # ── ISBN-13 校验 ──
    # 计算前 12 位的加权和（奇数位权重 1，偶数位权重 3）
    s = 0
    i = 0
    while i < 12:
        digit = int(isbn[i])
        if i % 2 == 0:             # 第 1,3,5... 位（索引 0,2,4...）：权重 1
            s += digit * 1
        else:                       # 第 2,4,6... 位（索引 1,3,5...）：权重 3
            s += digit * 3
        i += 1
    
    # 期望的校验位
    n = (10 - s % 10) % 10
    expected_check = str(n)
    actual_check = isbn[12]
    
    if actual_check == expected_check:
        print("✅ 合法的 ISBN-13")
    else:
        print(f"❌ 非法的 ISBN-13（期望校验位：'{expected_check}'，实际：'{actual_check}'）")

else:
    print("❌ 非法的 ISBN（长度必须为 10 或 13 位）")
```

```python
# ====== ISBN 校验 — 批量验证课本示例 ======
# 参考答案（验证用）：

def check_isbn(isbn):
    """返回校验结果字符串"""
    if len(isbn) == 10:
        s = 0
        i = 0
        while i < 9:
            s += int(isbn[i]) * (10 - i)
            i += 1
        n = (11 - s % 11) % 11
        expected = 'X' if n == 10 else str(n)
        return "Valid ISBN-10" if isbn[9] == expected else "Not a valid ISBN-10"
    elif len(isbn) == 13:
        s = 0
        i = 0
        while i < 12:
            digit = int(isbn[i])
            s += digit * (1 if i % 2 == 0 else 3)
            i += 1
        n = (10 - s % 10) % 10
        return "Valid ISBN-13" if isbn[12] == str(n) else "Not a valid ISBN-13"
    else:
        return "Not a valid ISBN"

# 课本示例
test_isbns = [
    "020151141",     # 9位 → 不合法
    "020151141X",    # 10位 → 合法
    "0306406153",    # 10位 → 不合法
    "0306406152",    # 10位 → 合法
    "9782340076556", # 13位 → 合法
]

print("ISBN 批量验证（课本示例）：")
print("-" * 40)
for isbn in test_isbns:
    result = check_isbn(isbn)
    icon = "✅" if "Valid" in result else "❌"
    print(f"  {isbn:15s} → {icon} {result}")
```
