## 第一章：引言 — 什么是命令行？

### 1.1 图形界面 vs 命令行

| | 图形界面 (GUI) | 命令行 (CLI) |
|:---|:---|:---|
| 操作方式 | 鼠标点击、拖拽 | 键盘输入文本指令 |
| 效率 | 直观但慢 | 学习曲线陡，但熟练后极快 |
| 批量处理 | 繁琐（逐个文件操作） | 一行命令搞定 |
| 自动化 | 困难 | 天然支持脚本 |
| 远程操作 | 需要远程桌面 | SSH 一行连接 |

> 🎯 命令行不是"过时技术"——它是程序员、运维工程师、数据科学家的**核心生产力工具**。

### 1.2 三种 Shell 家族

在 Windows 上，你有三种命令行环境可选：

| Shell | 可执行文件 | 特点 | 适合场景 |
|:---|:---|:---|:---|
| **CMD** | `cmd.exe` | Windows 原生，历史悠久，语法简单 | 基础文件操作、简单批处理 |
| **PowerShell** | `powershell.exe` | 微软现代 Shell，面向对象，功能强大 | 系统管理、自动化运维、.NET 集成 |
| **Bash** | `bash.exe`（Git Bash） | Unix/Linux 标准 Shell，开源生态 | 跨平台开发、Git 操作、开源工具链 |

> 💡 **三者可以共存，各取所长。** 日常开发中，PowerShell 和 Bash 最常用。

### 1.3 环境自检

运行下面的代码，确认你的系统上三种 Shell 是否可用：

````python
# ====== 环境自检：确认三种 Shell 是否可用 ======
import subprocess

def check_shell(name, cmd):
    """检查某个 Shell 是否在 PATH 中"""
    try:
        result = subprocess.run(cmd, capture_output=True, text=True, shell=True, timeout=5)
        version = result.stdout.strip().split('\n')[0] if result.stdout else "OK"
        return f"✅ {name} — 可用 ({version[:60]})"
    except Exception as e:
        return f"❌ {name} — 不可用: {e}"

print("=" * 50)
print("  Windows 命令行环境检测")
print("=" * 50)
print()

# 检查 CMD
print(check_shell("CMD", "cmd /c ver"))

# 检查 PowerShell
print(check_shell("PowerShell", 'powershell -NoProfile -Command "$PSVersionTable.PSVersion"'))

# 检查 Git Bash
print(check_shell("Bash (Git Bash)", 'bash -c "bash --version"'))

# 检查 Git（Git Bash 依赖）
print(check_shell("Git", "git --version"))

print()
print("=" * 50)
print("  如果三项都是 ✅，你可以继续学习了！")
print("=" * 50)
````

## 第二章：各解释器特点与基本命令

### 2.1 三种 Shell 的"第一印象"

先来看看同一个操作（查看当前目录），三种 Shell 分别怎么写：

````python
# ====== 同一操作，三种 Shell 写法对比 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

print("┌──────────────────────────────────────────────────┐")
print("│   同一操作：查看当前目录内容                       │")
print("├──────────────────────────────────────────────────┤")

# CMD 方式
print("│  CMD     (dir)                                   │")
r1 = subprocess.run(['cmd', '/c', f'cd /d {BASE_DIR} && dir /b'], 
                     capture_output=True, text=True, shell=True)
for line in r1.stdout.strip().split('\n'):
    print(f"│    {line}")

print("├──────────────────────────────────────────────────┤")

# PowerShell 方式
print("│  PowerShell (Get-ChildItem)                      │")
r2 = subprocess.run(['powershell', '-NoProfile', '-Command', 
                      f'Set-Location "{BASE_DIR}"; Get-ChildItem | Select-Object -ExpandProperty Name'],
                     capture_output=True, text=True, shell=True)
for line in r2.stdout.strip().split('\n'):
    print(f"│    {line}")

print("├──────────────────────────────────────────────────┤")

# Bash 方式
print("│  Bash    (ls)                                    │")
r3 = subprocess.run(['bash', '-c', f'cd "{BASE_DIR}" && ls'],
                     capture_output=True, text=True)
for line in r3.stdout.strip().split('\n'):
    print(f"│    {line}")

print("└──────────────────────────────────────────────────┘")
````

### 2.2 CMD 命令提示符 — 基础命令速查

CMD (cmd.exe) 是 Windows 最古老的命令行解释器。语法简单，适合快速操作。

| 命令 | 用途 | 示例 |
|:---|:---|:---|
| `dir` | 列出目录内容 | `dir /b`（简洁模式） |
| `cd` | 切换目录 | `cd C:\Users` |
| `md` / `mkdir` | 创建目录 | `md new_folder` |
| `rd` / `rmdir` | 删除目录 | `rd /s folder` |
| `copy` | 复制文件 | `copy a.txt b.txt` |
| `move` | 移动文件 | `move a.txt ..\` |
| `del` | 删除文件 | `del *.tmp` |
| `type` | 显示文件内容 | `type readme.txt` |
| `echo` | 输出文本 | `echo Hello` |
| `cls` | 清屏 | `cls` |

````python
# ====== CMD 基础命令演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

def run_cmd(command, description=""):
    """运行 CMD 命令并打印结果"""
    if description:
        print(f"\n{'='*50}")
        print(f"  {description}")
        print(f"{'='*50}")
    full_cmd = f'cd /d "{BASE_DIR}" && {command}'
    print(f"  CMD> {command}")
    result = subprocess.run(['cmd', '/c', full_cmd], capture_output=True, text=True, shell=True)
    output = result.stdout.strip()
    if output:
        for line in output.split('\n')[:15]:  # 最多显示15行
            print(f"       {line}")
    if result.stderr.strip():
        print(f"  [stderr: {result.stderr.strip()[:100]}]")

# 1. dir — 列出目录
run_cmd('dir /b', 'dir /b — 列出当前目录（简洁模式）')

# 2. echo — 输出文本
run_cmd('echo Hello CMD!', 'echo — 输出文本')

# 3. type — 查看文件内容
run_cmd('type readme.txt', 'type — 查看文件内容')

# 4. cd — 切换目录
run_cmd('cd data && dir /b && cd ..', 'cd — 切换目录并列出内容')

# 5. md — 创建目录
run_cmd('md temp_folder 2>nul && dir /b', 'md — 创建目录')
````

### 2.3 PowerShell — 基础命令速查

PowerShell 的命令叫 **Cmdlet**（发音：command-let），格式统一为 `动词-名词`。

| Cmdlet | 别名 | 用途 |
|:---|:---|:---|
| `Get-ChildItem` | `ls`, `dir` | 列出目录内容 |
| `Set-Location` | `cd` | 切换目录 |
| `Get-Content` | `cat`, `type` | 查看文件内容 |
| `New-Item` | `mkdir`, `ni` | 创建文件/目录 |
| `Remove-Item` | `rm`, `del` | 删除文件/目录 |
| `Copy-Item` | `cp`, `copy` | 复制 |
| `Move-Item` | `mv`, `move` | 移动 |
| `Write-Output` | `echo`, `write` | 输出 |
| `Get-Process` | `ps` | 查看进程 |
| `Get-Service` | — | 查看服务 |

> 💡 PowerShell 兼容大部分 CMD 命令和 Bash 别名（如 `ls`、`cat`、`rm`），方便迁移。

````python
# ====== PowerShell 基础命令演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

def run_ps(command, description=""):
    """运行 PowerShell 命令并打印结果"""
    if description:
        print(f"\n{'='*50}")
        print(f"  {description}")
        print(f"{'='*50}")
    full_cmd = f'Set-Location "{BASE_DIR}"; {command}'
    print(f"  PS> {command}")
    result = subprocess.run(['powershell', '-NoProfile', '-Command', full_cmd],
                           capture_output=True, text=True, shell=True)
    output = result.stdout.strip()
    if output:
        for line in output.split('\n')[:15]:
            print(f"       {line}")
    if result.stderr.strip():
        print(f"  [错误: {result.stderr.strip()[:150]}]")

# 1. Get-ChildItem
run_ps('Get-ChildItem | Select-Object Name, Length', 'Get-ChildItem — 列出目录（带属性）')

# 2. Write-Output
run_ps('Write-Output "Hello PowerShell!"', 'Write-Output — 输出文本')

# 3. Get-Content
run_ps('Get-Content readme.txt', 'Get-Content — 查看文件内容')

# 4. 别名也能用
run_ps('ls', '别名 ls — 等价于 Get-ChildItem')
````

### 2.4 Git Bash — 基础命令速查

Git Bash 提供了 Unix/Linux 风格命令行体验，是跨平台开发者的首选。

| 命令 | 用途 | 示例 |
|:---|:---|:---|
| `ls` | 列出目录 | `ls -la` |
| `cd` | 切换目录 | `cd /c/Users` |
| `pwd` | 显示当前路径 | `pwd` |
| `mkdir` | 创建目录 | `mkdir newdir` |
| `rm` | 删除文件/目录 | `rm -rf folder` |
| `cp` | 复制 | `cp a.txt b.txt` |
| `mv` | 移动/重命名 | `mv old.txt new.txt` |
| `cat` | 查看文件 | `cat readme.txt` |
| `echo` | 输出文本 | `echo Hello` |
| `clear` | 清屏 | `clear` |
| `touch` | 创建空文件 | `touch new.txt` |

````python
# ====== Git Bash 基础命令演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

def run_bash(command, description=""):
    """运行 Bash 命令并打印结果"""
    if description:
        print(f"\n{'='*50}")
        print(f"  {description}")
        print(f"{'='*50}")
    full_cmd = f'cd "{BASE_DIR}" && {command}'
    print(f"  $ {command}")
    result = subprocess.run(['bash', '-c', full_cmd], capture_output=True, text=True)
    output = result.stdout.strip()
    if output:
        for line in output.split('\n')[:15]:
            print(f"       {line}")
    if result.stderr.strip():
        print(f"  [错误: {result.stderr.strip()[:150]}]")

# 1. ls -la
run_bash('ls -la', 'ls -la — 列出目录（详细信息）')

# 2. pwd
run_bash('pwd', 'pwd — 显示当前路径')

# 3. echo
run_bash('echo "Hello from Bash!"', 'echo — 输出文本')

# 4. cat
run_bash('cat readme.txt', 'cat — 查看文件内容')

# 5. touch + ls
run_bash('touch newfile.txt && ls *.txt', 'touch — 创建空文件 + ls 通配符')
````

## 第三章：文件系统操作

### 3.1 CMD — 文件与目录操作

````python
# ====== CMD 文件操作深入演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

def run_cmd_multi(commands):
    """运行多条 CMD 命令"""
    full = f'cd /d "{BASE_DIR}" && ' + ' && '.join(commands)
    result = subprocess.run(['cmd', '/c', full], capture_output=True, text=True, shell=True)
    print(result.stdout)

print("=== 1. 创建多级目录 ===")
print("CMD> md level1\\level2\\level3")
run_cmd_multi(['md level1\\level2\\level3 2>nul', 'dir /b /s level1'])

print("\n=== 2. 创建文件并写入内容 ===")
print("CMD> echo 你好世界 > hello.txt")
run_cmd_multi(['echo 你好世界 > hello.txt', 'type hello.txt'])

print("\n=== 3. 追加内容到文件 ===")
print('CMD> echo 第二行 >> hello.txt')
run_cmd_multi(['echo 第二行 >> hello.txt', 'type hello.txt'])

print("\n=== 4. 复制文件 ===")
print("CMD> copy hello.txt hello_backup.txt")
run_cmd_multi(['copy hello.txt hello_backup.txt >nul', 'dir /b hello*'])

print("\n=== 5. 重命名文件 ===")
print("CMD> ren hello_backup.txt hello_old.txt")
run_cmd_multi(['ren hello_backup.txt hello_old.txt', 'dir /b hello*'])

print("\n=== 6. 删除文件和目录 ===")
print("CMD> del hello_old.txt && rd /s /q level1")
run_cmd_multi(['del hello_old.txt', 'rd /s /q level1', 'dir /b'])
````

### 3.2 PowerShell — 文件与目录操作

PowerShell 的独特优势：**操作的是对象，不是文本**。

````python
# ====== PowerShell 文件操作深入演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

def run_ps_multi(commands):
    """运行多条 PowerShell 命令"""
    full = f'Set-Location "{BASE_DIR}"; ' + '; '.join(commands)
    result = subprocess.run(['powershell', '-NoProfile', '-Command', full],
                           capture_output=True, text=True, shell=True)
    print(result.stdout)

print("=== 1. 创建目录（一行创建多级） ===")
print("PS> New-Item -ItemType Directory -Path level1/level2/level3 -Force")
run_ps_multi([
    'New-Item -ItemType Directory -Path level1/level2/level3 -Force >$null',
    'Get-ChildItem level1 -Recurse | Select-Object FullName'
])

print("\n=== 2. 创建文件并写入内容 ===")
print("PS> Set-Content -Path hello.txt -Value '你好世界'")
run_ps_multi([
    "Set-Content -Path hello.txt -Value '你好世界'",
    'Get-Content hello.txt'
])

print("\n=== 3. 追加内容 ===")
print("PS> Add-Content -Path hello.txt -Value '第二行'")
run_ps_multi([
    "Add-Content -Path hello.txt -Value '第二行'",
    'Get-Content hello.txt'
])

print("\n=== 4. 复制 + 移动 + 删除 ===")
print("PS> Copy-Item / Move-Item / Remove-Item")
run_ps_multi([
    'Copy-Item hello.txt hello_copy.txt',
    'Move-Item hello_copy.txt hello_moved.txt -Force',
    'Get-ChildItem hello* | Select-Object Name',
    'Remove-Item hello_moved.txt, hello.txt -Force',
    'Remove-Item level1 -Recurse -Force',
    'Write-Output "清理完成！"',
    'Get-ChildItem'
])
````

### 3.3 Bash — 文件与目录操作

Bash 在文件操作上最为简洁，通配符和花括号展开是亮点。

````python
# ====== Bash 文件操作深入演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

def run_bash_multi(commands):
    """运行多条 Bash 命令"""
    full = f'cd "{BASE_DIR}" && ' + ' && '.join(commands)
    result = subprocess.run(['bash', '-c', full], capture_output=True, text=True)
    print(result.stdout)

print("=== 1. 花括号展开：一次创建多个目录 ===")
print("$ mkdir -p project/{src,test,docs}")
run_bash_multi([
    'mkdir -p project/{src,test,docs}',
    'ls -R project'
])

print("\n=== 2. 通配符操作 ===")
print('$ touch file{1..5}.txt  （花括号展开创建5个文件）')
run_bash_multi([
    'touch file{1..5}.txt',
    'ls *.txt'
])

print("\n=== 3. 复制、移动、删除 ===")
print("$ cp file1.txt file1_copy.txt && mv file2.txt file2_renamed.txt")
run_bash_multi([
    'cp file1.txt file1_copy.txt',
    'mv file2.txt file2_renamed.txt',
    'ls file*',
    'rm file*.txt project -rf',
    'echo "清理完成"',
    'ls'
])
````

## 第四章：文本处理与查看

### 4.1 CMD — find 和 findstr

CMD 提供了 `find` 和 `findstr` 两个文本搜索工具。

````python
# ====== CMD 文本处理：find 和 findstr ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

# 先确认我们的示例文件存在
print("=== 查看 data/log.txt 内容 ===")
r = subprocess.run(['cmd', '/c', f'cd /d "{BASE_DIR}" && type data\\log.txt'], 
                   capture_output=True, text=True, shell=True)
print(r.stdout)

print("\n=== find — 搜索包含 ERROR 的行（计行数）===")
r = subprocess.run(['cmd', '/c', f'cd /d "{BASE_DIR}" && find /c "ERROR" data\\log.txt'], 
                   capture_output=True, text=True, shell=True)
print(r.stdout)

print("\n=== findstr — 正则搜索（支持正则表达式）===")
r = subprocess.run(['cmd', '/c', f'cd /d "{BASE_DIR}" && findstr /i "ERROR|WARN" data\\log.txt'], 
                   capture_output=True, text=True, shell=True)
print(r.stdout)
````

### 4.2 PowerShell — Select-String（grep 等价物）

## 第六章：进阶用法 — 管道、重定向、命令组合

### 6.1 理解管道（Pipeline）

**管道** 把一个命令的**输出**作为下一个命令的**输入**。

```
命令A → | → 命令B → | → 命令C → 屏幕
```

| Shell | 管道符号 | 传递内容 |
|:---|:---|:---|
| CMD | `|` | 纯文本 |
| PowerShell | `|` | **对象**（这是最强大的地方！） |
| Bash | `|` | 纯文本 |

````python
# ====== PowerShell 管道之美：面向对象 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

print("=" * 60)
print("  PowerShell 管道：传递对象，不只是文本")
print("=" * 60)

# PowerShell 管道链：获取文件 → 过滤 → 排序 → 选择属性 → 格式化
r = subprocess.run(['powershell', '-NoProfile', '-Command', f'''
Set-Location "{BASE_DIR}"
Get-ChildItem -Recurse |
    Where-Object {{ $_.Length -gt 0 }} |
    Sort-Object Length -Descending |
    Select-Object Name, @{{N="Size";E={{$_.Length}}}}, @{{N="Path";E={{$_.Directory.Name}}}} |
    Format-Table -AutoSize
'''], capture_output=True, text=True, shell=True)
print(r.stdout)
print("这条管道做了：获取所有文件 → 过滤非空 → 按大小排序 → 选属性 → 表格输出")
````

### 6.2 Bash 管道链实战

Bash 的管道虽然传递的是文本，但配合 grep/sed/awk/sort/uniq 的组合非常灵活。

````python
# ====== Bash 管道链实战 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

# 复杂管道：统计日志中各等级的数量
print("=== Bash 管道链：日志等级统计 ===")
print("$ cat log.txt | grep -oP '\\[.*?\\] \\[\\K\\w+' | sort | uniq -c | sort -rn\n")
r = subprocess.run(['bash', '-c', f'''
cd "{BASE_DIR}"
echo "=== 原始日志（前5行）==="
head -5 data/log.txt
echo ""
echo "=== 管道处理：统计各日志等级数量 ==="
cat data/log.txt | grep -oP "\\[.*?\\] \\[\\K\\w+" | sort | uniq -c | sort -rn
'''], capture_output=True, text=True)
print(r.stdout)

print("管道拆解：")
print("  ① cat → 读取文件")
print("  ② grep -oP → 用正则提取日志等级（INFO/WARN/ERROR）")
print("  ③ sort → 排序（uniq 要求先排序）")
print("  ④ uniq -c → 统计每个值出现次数")
print("  ⑤ sort -rn → 按次数倒序排列")
````

### 6.3 重定向

**重定向** 把命令的输出保存到文件（`>`覆盖，`>>`追加）或从文件读取输入（`<`）。

| 符号 | 含义 | 示例 |
|:---|:---|:---|
| `>` | 输出到文件（覆盖） | `echo hello > file.txt` |
| `>>` | 输出到文件（追加） | `echo world >> file.txt` |
| `<` | 从文件读取输入 | `sort < data.txt` |
| `2>` | 错误输出重定向 | `cmd 2> error.log` |
| `>&2` | 输出到 stderr | `echo error >&2` |
| `/dev/null` | 丢弃输出 | `cmd > /dev/null 2>&1` |

````python
# ====== 重定向演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

# Bash 重定向演示
r = subprocess.run(['bash', '-c', f'''
cd "{BASE_DIR}"

echo "=== 1. 输出重定向：> 覆盖写入 ==="
echo "第一行内容" > redirect_demo.txt
cat redirect_demo.txt

echo ""
echo "=== 2. 追加：>> 追加到文件末尾 ==="
echo "第二行内容" >> redirect_demo.txt
echo "第三行内容" >> redirect_demo.txt
cat redirect_demo.txt

echo ""
echo "=== 3. 输入重定向：< 把文件内容喂给命令 ==="
wc -l < redirect_demo.txt

echo ""
echo "=== 4. 同时重定向 stdout 和 stderr ==="
ls existing_file.txt non_exist.txt > output.txt 2>&1
cat output.txt

echo ""
echo "=== 清理 ==="
rm redirect_demo.txt output.txt
echo "清理完成"
'''], capture_output=True, text=True)
print(r.stdout)
````

### 6.4 命令连接符

| 符号 | CMD | PowerShell | Bash | 含义 |
|:---|:---|:---|:---|:---|
| `&&` | ✅ | ❌（用 `;`） | ✅ | 前一个成功才执行下一个 |
| `\|\|` | ✅ | ❌ | ✅ | 前一个失败才执行下一个 |
| `;` | ❌（用 `&`） | ✅ | ✅ | 无论成败都执行下一个 |
| `&` | ✅（单&） | ❌ | ✅（后台运行） | CMD：无条件连接；Bash：后台 |

````python
# ====== 命令连接符演示 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

print("=== CMD: &&（成功才继续）===")
r = subprocess.run(['cmd', '/c', f'cd /d "{BASE_DIR}" && echo 第一步成功 && echo 第二步也成功'], 
                   capture_output=True, text=True, shell=True)
print(r.stdout)

print("=== CMD: ||（失败时执行）===")
r = subprocess.run(['cmd', '/c', f'cd /d "{BASE_DIR}" && dir 不存在的文件 >nul 2>&1 || echo 文件不存在！'], 
                   capture_output=True, text=True, shell=True)
print(r.stdout)

print("=== Bash: && 和 || 的组合 ===")
r = subprocess.run(['bash', '-c', f'''
cd "{BASE_DIR}"
echo "--- 文件存在时 ---"
test -f readme.txt && echo "readme.txt 存在" || echo "readme.txt 不存在"
echo "--- 文件不存在时 ---"
test -f ghost.txt && echo "ghost.txt 存在" || echo "ghost.txt 不存在"
'''], capture_output=True, text=True)
print(r.stdout)
````

### 7.1 案例一：文件批量重命名

**场景**：把当前目录下所有 `.txt` 文件重命名为 `.md` 文件。

````python
# ====== 批量重命名：三 Shell 对比 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')
BASH_BASE = BASE_DIR.replace('\\', '/')

# 准备：创建一些 .txt 文件
os.makedirs(os.path.join(BASE_DIR, 'rename_test'), exist_ok=True)
for i in range(1, 6):
    with open(os.path.join(BASE_DIR, 'rename_test', f'doc{i}.txt'), 'w') as f:
        f.write(f'Content {i}')

print("=== 重命名前 ===")
r = subprocess.run(['cmd', '/c', f'dir /b "{BASE_DIR}\\rename_test"'], capture_output=True, text=True, shell=True)
print(r.stdout)

# CMD 方式
print("=== CMD: 批量重命名（用 for 循环）===")
r = subprocess.run(['cmd', '/c', f'cd /d "{BASE_DIR}\\rename_test" && for %f in (*.txt) do ren "%f" "%~nf.md" && dir /b'], 
                   capture_output=True, text=True, shell=True)
print(r.stdout[:300])

# 恢复 .txt 用于下一个演示
for fn in os.listdir(os.path.join(BASE_DIR, 'rename_test')):
    if fn.endswith('.md'):
        old = os.path.join(BASE_DIR, 'rename_test', fn)
        new = os.path.join(BASE_DIR, 'rename_test', fn.replace('.md', '.txt'))
        os.rename(old, new)

# PowerShell 方式
print("=== PowerShell: 批量重命名（一行搞定）===")
r = subprocess.run(['powershell', '-NoProfile', '-Command',
    f'Set-Location "{BASE_DIR}\\rename_test"; Get-ChildItem *.txt | Rename-Item -NewName {{$_.Name -replace "\\.txt$", ".md"}}; Get-ChildItem'],
    capture_output=True, text=True, shell=True)
print(r.stdout[:300])

# 恢复
for fn in os.listdir(os.path.join(BASE_DIR, 'rename_test')):
    if fn.endswith('.md'):
        old = os.path.join(BASE_DIR, 'rename_test', fn)
        new = os.path.join(BASE_DIR, 'rename_test', fn.replace('.md', '.txt'))
        os.rename(old, new)

# Bash 方式
print("=== Bash: 批量重命名（最简洁）===")
r = subprocess.run(['bash', '-c', f'''
cd "{BASH_BASE}/rename_test"
for f in *.txt; do mv "$f" "${{f%.txt}}.md"; done
ls
'''], capture_output=True, text=True)
print(r.stdout[:300])

# 清理
import shutil
shutil.rmtree(os.path.join(BASE_DIR, 'rename_test'), ignore_errors=True)
print("✅ 清理完成")
````

````python
# ====== PowerShell 综合：大文件扫描 ======
import subprocess, os

BASE_DIR = os.path.join(os.environ['TEMP'], 'cli_tutorial')

r = subprocess.run(['powershell', '-NoProfile', '-Command', f'''
Set-Location "{BASE_DIR}"

Write-Host "╔══════════════════════════════════════╗" -ForegroundColor Cyan
Write-Host "║   📁 大文件扫描器（> 10 字节）        ║" -ForegroundColor Cyan
Write-Host "╚══════════════════════════════════════╝" -ForegroundColor Cyan
Write-Host ""

# 获取所有文件，过滤大小 > 10 字节，排序，前10
$files = Get-ChildItem -Recurse -File |
    Where-Object {{ $_.Length -gt 10 }} |
    Sort-Object Length -Descending |
    Select-Object -First 10

# 管道输出到 Format-Table
$files | Select-Object @{{N="文件";E={{$_.Name}}}},
                        @{{N="大小(B)";E={{$_.Length}}}},
                        @{{N="目录";E={{$_.Directory.Name}}}} |
    Format-Table -AutoSize

Write-Host "共找到 $(($files | Measure-Object).Count) 个大于 10 字节的文件" -ForegroundColor Green
'''], capture_output=True, text=True, shell=True)
print(r.stdout)
````

````python
# ====== CMD 高级批处理：延时与循环 ======
import subprocess

print("=" * 50)
print("  CMD 高级技巧：超时等待 + 循环重试")
print("=" * 50)

# 模拟：ping 服务器直到成功（最多5次）
print("CMD> for /L %i in (1,1,5) do @ping -n 1 8.8.8.8 >nul && echo 第%i次: 成功 && goto :end || echo 第%i次: 失败")
print("(这是 CMD 中罕见的类似 while 循环的用法)")

# 实际演示：用 ping 127.0.0.1 模拟（一定成功的）
r = subprocess.run(['cmd', '/c', 
    'for /L %i in (1,1,3) do @echo 尝试 %i... && ping -n 1 127.0.0.1 >nul && echo ✅ 成功'],
    capture_output=True, text=True, shell=True)
print("\n实际输出：")
print(r.stdout)

print("\nCMD 延时技巧：")
print("  timeout /t 5          — 等待5秒（可按任意键跳过）")
print("  timeout /t 5 /nobreak — 等待5秒（不可跳过）")
print("  ping -n 6 127.0.0.1 >nul — 近似延时5秒的老技巧")
````
