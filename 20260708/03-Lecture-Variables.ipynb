## 第一部分：变量 (Variables)

### 什么是变量？

**变量** 就像你书桌上的一个 **收纳盒**：
- 你给盒子贴一个标签（变量名）
- 往盒子里放东西（赋值）
- 之后只要喊标签名字，就能拿到里面的东西

```python
盒子标签 = "里面的东西"
```

在 Python 里，创建一个变量只需要一行代码：

````python
# 创建一个变量 name，里面放字符串 "小明"
name = "小明"

# 创建一个变量 age，里面放数字 12
age = 12

# 打印看看里面有什么
print(name)
print(age)
````

**关键概念**：
- `=` 不是数学里的"等于号"，而是 **赋值号** — 把右边的值"装进"左边的变量
- 变量名在左边，值在右边

````python
# 你可以在变量里放各种类型的数据
my_string = "你好"      # 字符串（文字）
my_int = 42             # 整数
my_float = 3.14         # 小数（浮点数）
my_bool = True          # 布尔值（True/False）

print(my_string)
print(my_int)
print(my_float)
print(my_bool)
````

### 变量命名规则

给变量取名就像给你的宠物取名 — 有些名字可以，有些不行：

| 规则 | ✅ 允许 | ❌ 不允许 |
|:---|:---|:---|
| 长度 | 多长都行 | — |
| 字符 | 字母、数字、下划线 `_` | 空格、`$`、`@`、`#` 等特殊符号 |
| 第一个字符 | **不能**是数字 | `3foo` ❌ |
| 大小写 | 区分大小写 | `foo` 和 `Foo` 是两个不同变量 |
| 关键字 | — | `print`、`if`、`while` 等 Python 专用词不能用 |

````python
# 合法的变量名
_foo = "下划线开头也可以"
foo = "小写"
FoO = "大小写不同"
foobar123 = "数字可以在后面"
my_very_long_variable_name_that_is_hard_to_read = "多长都行"

print(_foo)
print(foo)
print(FoO)
print(foobar123)
print(my_very_long_variable_name_that_is_hard_to_read)
````

### 变量是怎么存在电脑里的？（了解一下）

每个变量在电脑内存里都有一个**地址**，就像每个房子有一个门牌号。
可以用 `id()` 函数查看变量的"门牌号"：

````python
# id() 返回变量在内存中的地址
var1 = 100
var2 = var1  # var2 指向跟 var1 同一个值
var3 = 100   # var3 也指向 100

print("var1 的内存地址:", id(var1))
print("var2 的内存地址:", id(var2))
print("var3 的内存地址:", id(var3))

# 注意：三个地址一样！因为 Python 发现 100 已经存在，就重复用了
````

````python
# 但是如果你改变 var1 的值，它的地址会变
var1 = 1000
print("var1 的新地址:", id(var1))
print("var2 的地址没变:", id(var2))

# var1 现在指向新的值 1000，但 var2 仍然指向 100
````

### 一行声明多个变量（多重赋值）

Python 有一种酷炫的写法：一口气创建多个变量。左边写变量名，右边写值，用逗号隔开。

````python
# 一行创建三个变量
a, b, c = 3, "Hello", 7

print(a)
print(b)
print(c)
````

### 🔄 交换两个变量的值

假设 `a=3`、`b=21`，想交换成 `a=21`、`b=3`。

在其他语言里需要一个"临时盒子"（`temp=a; a=b; b=temp`），但 Python 一行搞定：

````python
a = 3
b = 21
print("交换前：a =", a, "b =", b)

# Python 的酷炫交换
a, b = b, a
print("交换后：a =", a, "b =", b)
````

**为什么能这么写？**
- 右边 `b, a` 先算出两个值 `(21, 3)`
- 左边 `a, b` 依次接收：`a=21`, `b=3`
- 相当于多重赋值的妙用

## 第二部分：用户输入 input()

### 怎么让程序"听"你说话？

`input()` 是 Python 的内置函数，它会在终端**等待你打字**，你按回车后，它把你打的字**作为字符串返回**。

````python
# input() 最简单的用法
# 运行后，在下面的输入框里打字，按回车
input("请随便打几个字然后回车：")
````

`input()` 返回的值**一定要存到变量里**，否则就丢了！

````python
# 把输入存到变量里
user_name = input("你叫什么名字？ ")

# 然后可以反复使用
print("你好，" + user_name + "！")
print("欢迎你，" + user_name + "！")
````

### ⚠️ 重要：input() 返回的永远是字符串

即使你输入的是数字，`input()` 也把它当**文字**处理。看下面的坑：

````python
# 试试输入一个数字，比如 12
age = input("你几岁了？ ")
print("你输入的是：", age)
print("它的类型是：", type(age))

# 注意 type() 显示的是 <class 'str'>，是字符串，不是数字！
````

### 类型转换：把字符串变成数字

如果需要做数学运算，得用 `int()` 或 `float()` 转换：

````python
# 输入出生年份
birth_year = input("你是哪一年出生的？ ")

# 不转换会怎样？
# 下面会报错！字符串不能和数字做运算
# print("直接用字符串加：", birth_year + 2026 - 2026)
````

````python
# 正确做法：用 int() 转换
birth_year = int(input("你是哪一年出生的？ "))
age = 2026 - birth_year
print("你今年", age, "岁！")
````

### 也可以嵌套着写：省掉中间变量

```python

```

````python
# 嵌套写法：input 的结果直接传给 int()
height = float(input("你的身高（米）："))
print("你的身高是", height, "米")
print("换算成厘米是", height * 100, "厘米")
````

## 第三部分：print 的进阶用法

### 输出多个值

`print()` 可以一次接收**多个参数**，用逗号隔开，输出时默认用空格分隔：

````python
# 一次打印多个东西
print("苹果", "香蕉", "橙子", "葡萄")

# 混合文字和变量
name = "小红"
score = 95
print("姓名：", name, "成绩：", score)
````

### sep 参数：控制分隔符

默认是空格隔开，但你可以改成任何东西：

````python
# sep 改变分隔符
print("a", "b", "c", sep=" >>> ")
print("2026", "07", "08", sep="-")
print("apple", "banana", "cherry", sep="|")
````

### end 参数：控制结尾

默认 `print()` 结尾会**换行**（`end="\n"`）。你可以改成别的：

````python
# print 默认结尾是换行
print("第一行")
print("第二行")
````

````python
# 把结尾改成 >>>，就不会换行了
print(1, end=" >>> ")
print(2, end=" >>> ")
print(3, end=" >>> ")
print("发射！")
````

````python
# sep 和 end 可以一起用
print("倒计时", 5, 4, 3, 2, 1, sep="...", end="!!!")
print(" 新年快乐！")
````

## 第四部分：字符串格式化 format()

### 把变量放进字符串里

用 `+` 拼接字符串太麻烦，`format()` 方法可以优雅地插入变量：

````python
# format 基础用法：用 {} 占位
name = "小明"
height = 1.65

print("我叫{}，身高{}米".format(name, height))

# {0} 表示第0个参数，{1} 表示第1个参数
print("身高{1}米，我叫{0}".format(name, height))
````

### 对齐与填充（视觉排版）

| 符号 | 意思 |
|:---:|:---|
| `<` | 左对齐 |
| `>` | 右对齐 |
| `^` | 居中对齐 |

格式：`{:<10}` 表示占10个格子，左对齐

````python
# 字符串对齐
print("|{:>10}|".format("右对齐"))     # 右对齐，占10格
print("|{:<10}|".format("左对齐"))     # 左对齐
print("|{:^10}|".format("居中"))       # 居中
print("|{:_>10}|".format("右对齐"))    # 右对齐，用 _ 填充空白
print("|{:-^10}|".format("居中"))      # 居中，用 - 填充空白
````

### 截断字符串

`{:.4}` 表示只取前4个字符：

````python
# 截断字符串
word = "promethee"
print("{:.4}".format(word))           # 前4个字符
print("{:>10.5}".format(word))        # 右对齐10格，取前5个
print("{:-^10.2}".format(word))       # 居中10格，取前2个，-填充
````

### 数字格式化

控制数字的输出格式：

````python
# 整数格式化
print("{:d}".format(123))         # 整数
print("{:010d}".format(123))      # 占10位，前面补0

# 小数格式化
print("{:f}".format(1.23456789)) # 默认6位小数
print("{:.2f}".format(1.23456789)) # 保留2位小数
print("{:010.3f}".format(1.23456789)) # 总长10位，3位小数
````

````python
# 百分比格式
print("你的正确率是 {:.2%}".format(0.85714))

# 二进制
print("数字 42 的二进制是 {:b}".format(42))

# 字符（ASCII码）
print("ASCII 65 对应字符 {:c}".format(65))
````

### ✨ 更简单的方式：f-string（Python 3.6+）

f-string 是现在最推荐的方法，直接在字符串前加 `f`，变量名写在大括号里：

````python
# f-string 更直观
name = "小明"
age = 12
height = 1.55

print(f"我叫{name}，今年{age}岁，身高{height}米")
print(f"明年我就{age + 1}岁了")  # 里面甚至可以写表达式
print(f"身高保留2位小数：{height:.2f}")
print(f"|{name:>10}|")  # 对齐也能用
````
