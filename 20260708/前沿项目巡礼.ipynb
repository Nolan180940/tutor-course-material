# 🤖 项目一：OpenCode

> [github.com/anomalyco/opencode](https://github.com/anomalyco/opencode)

## 它是什么？

一个**开源的 AI 编程助手**，住在终端里。你可以把它理解成「免费的 Claude Code」。

```
你：帮我写一个贪吃蛇游戏
OpenCode：写好了，运行 snake.py 试试
你：加个分数显示
OpenCode：已添加，左上角显示当前分数
```

## 跟 Claude Code 什么关系？

| | Claude Code | OpenCode |
|:---|:---|:---|
| 开源？ | ❌ 闭源 | ✅ MIT 开源 |
| 用谁的模型 | 只能用 Claude | 支持几十种模型 |
| 价格 | 要付费 | 免费 |
| 怎么用 | 终端里聊天 | 终端里聊天 |

> 简单说：**OpenCode = 开源的 Claude Code**，功能差不多，但你可以用 DeepSeek、GLM 等国产模型驱动它。

## 技术栈

| TypeScript | 核心语言 |
| React | 桌面应用界面 |
| MCP 协议 | 连接外部工具 |

# 📋 项目二：PasteMD

> [github.com/RICHQAQ/PasteMD](https://github.com/RICHQAQ/PasteMD)

## 它是什么？

一个**常驻托盘的小工具**。你从 ChatGPT/DeepSeek 复制内容，按 `Ctrl+Shift+B`，它自动转成 Word 文档并贴在光标位置。

## 解决什么痛点？

```
写论文/报告时的痛苦：
1. 从 DeepSeek 复制 Markdown
2. 贴到 Word 里 — 全是乱码！
3. 公式显示成一堆 $ 符号
4. 手动一个个调格式 😭

有了 PasteMD：
1. 从 DeepSeek 复制 Markdown
2. 按 Ctrl+Shift+B
3. 完美格式出现在 Word 里 🎉
```

## 亮点功能

| 功能 | 说明 |
|:---|:---|
| 📝 Markdown → Word | 一键转换，格式完美 |
| 📊 表格 → Excel | 自动识别 Markdown 表格 |
| 🌐 网页 AI 回复 | 直接复制网页内容也能转 |
| 🧮 数学公式 | LaTeX 公式完美渲染 |
| 🎯 应用扩展 | 不同 app 用不同粘贴模式 |

## AI 兼容性

支持 ChatGPT、DeepSeek、Kimi、通义千问、豆包、智谱清言、Gemini、Grok、Claude……几乎全部主流 AI！

## 技术栈

| 技术 | 作用 |
|:---|:---|
| Python 3.12 | 核心语言 |
| Pandoc | Markdown → DOCX 转换引擎 |
| PyQt / Tkinter | 托盘图标 + 设置界面 |
| python-docx | Word 样式后处理 |

> [github.com/Nolan180940/md-to-word](https://github.com/Nolan180940/md-to-word)

## 它是什么？

你老师写的开源项目！一个**在线 + 桌面**的 Markdown 转 Word 工具，专注处理 LLM 生成的内容。

## 跟 PasteMD 有什么区别？

| | PasteMD | MD-to-Word |
|:---|:---|:---|
| 用法 | 托盘热键粘贴 | 在线网页 / 桌面应用 |
| 适用场景 | 快速粘贴到 Word | 预览 + 编辑 + 导出 |
| 智能修复 | 基础 | 7 种自动修复 |
| 样式增强 | ❌ | ✅ 代码块阴影 + 引用缩进线 |
| 免费 | ✅ | ✅ 无需登录 |

## 智能修复引擎

自动修复 AI 输出中的常见问题：

| 问题 | 修复 |
|:---|:---|
| 隐形字符 | 移除零宽空格 |
| `\[...\]` | → `$$...$$` |
| `$ x^2 $` | → `$x^2$` |
| `<sup>` 标签 | → `^...^` |
| 代码块不闭合 | 自动补 ` ``` ` |
| 公式块不闭合 | 自动补 `$$` |
| 引用块格式 | 确保前后空行 |

# 🔧 幕后英雄：Pandoc

> [github.com/jgm/pandoc](https://github.com/jgm/pandoc) — ⭐ 37,000+

## 它是什么？

Pandoc 是一个**万能文档转换器**。它能把 Markdown 变成 Word、PDF、HTML、LaTeX……几乎你能想到的任何格式。

```
Markdown ──→ Pandoc ──→ Word (.docx)
                         → PDF
                         → HTML
                         → PowerPoint (.pptx)
                         → LaTeX
                         → EPUB 电子书
                         → ……几十种格式
```

## 🛠️ 安装 Pandoc

### Windows

```powershell
# 方式一：直接下载安装包（推荐）
start "" "https://pandoc.org/installing.html"

# 方式二：用 winget
winget install --id JohnMacFarlane.Pandoc

# 方式三：用 choco
choco install pandoc
```

安装完验证：

```powershell
pandoc --version
```

### macOS

```bash
brew install pandoc
```

### Linux (Ubuntu/Debian)

```bash
sudo apt-get install pandoc
```

## 🚀 基本使用

Pandoc 是命令行工具，基本语法：

```bash
pandoc 输入文件 -o 输出文件
```

### 常用示例

```bash
# Markdown → Word
pandoc report.md -o report.docx

# Markdown → PDF（需要 LaTeX）
pandoc report.md -o report.pdf

# Markdown → HTML 网页
pandoc report.md -o report.html

# Markdown → PowerPoint 幻灯片
pandoc slides.md -o slides.pptx

# 直接读取粘贴板内容转换（macOS）
pbpaste | pandoc -f markdown -t docx -o output.docx
```

### 进阶：指定格式细节

```bash
# 用 reference.docx 控制输出样式
pandoc input.md --reference-doc=template.docx -o output.docx

# 支持 LaTeX 数学公式
pandoc input.md --katex -o output.html

# 生成目录
pandoc input.md --toc -o output.docx
```

## 为什么两个项目都用它？

| | PasteMD | MD-to-Word |
|:---|:---:|:---:|
| 转换引擎 | Pandoc | Pandoc |
| 怎么调用 | `pandoc` 命令行 | `pypandoc` Python 库 |
